```python
# =====================================================
# AI CODE GENERATION & DEBUGGING USING FLAN-T5
# Single Cell - Google Colab
# =====================================================

# Install required libraries
!pip -q install transformers torch sentencepiece accelerate

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

print("Loading Code Generation Model...")

# Load tokenizer and model directly
flan_tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
flan_model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")

# Function for generating code or fixing code
def generate_code(prompt, max_new_tokens=120):
    input_ids = flan_tokenizer(prompt, return_tensors="pt").input_ids
    outputs = flan_model.generate(
        input_ids,
        max_new_tokens=max_new_tokens,
        do_sample=False
    )
    return flan_tokenizer.decode(outputs[0], skip_special_tokens=True)

# =====================================================
# Example 1 : Code Generation
# =====================================================

prompt1 = """
Write a Python function to check whether a number is prime.
"""

print("=" * 60)
print("CODE GENERATION")
print("=" * 60)

generated_code = generate_code(prompt1)

print(generated_code)

# =====================================================
# Example 2 : Debugging
# =====================================================

buggy_code = """
The following Python function should return the factorial of n.
Fix the bug and return the corrected code.

def factorial(n):
    result = 0
    for i in range(1, n+1):
        result = result * i
    return result
"""

print("\n" + "=" * 60)
print("DEBUGGING")
print("=" * 60)

fixed_code = generate_code(
    buggy_code,
    max_new_tokens=150
)

print(fixed_code)
```
